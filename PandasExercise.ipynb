```python
df=pd.DataFrame(file)
```

```python
max_annual_kerala = df.loc[df['STATE_UT_NAME'] == 'KERALA', 'ANNUAL'].max()
print(f"Maximum Annual Rainfall for KERALA: {max_annual_kerala}")

if max_annual_kerala <= 4000:
    print("No district in KERALA has an annual rainfall greater than 4000.")
```

```python
max_annual_overall = df['ANNUAL'].max()
print(f"Overall Maximum Annual Rainfall: {max_annual_overall}")

highest_rainfall_entry = df.loc[df['ANNUAL'] == max_annual_overall]
display(highest_rainfall_entry['DISTRICT'])
```

```python
top_5_states_rainfall = df.groupby('STATE_UT_NAME')['ANNUAL'].sum().sort_values(ascending=False).head()
print(top_5_states_rainfall)
```

To drop multiple columns, you can use the `drop()` method, passing a list of column names to be removed. The `axis=1` argument specifies that you are dropping columns (not rows).

```python
# Example: Drop the 'Jan-Feb', 'Mar-May', and 'Oct-Dec' columns
columns_to_drop = ['Jan-Feb', 'Mar-May', 'Oct-Dec','Jun-Sep']
df_dropped = df.drop(columns=columns_to_drop, axis=1)

print("DataFrame after dropping columns:")
display(df_dropped.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate state-wise mean rainfall
state_mean_rainfall = df.groupby('STATE_UT_NAME')['ANNUAL'].mean().reset_index()

# Sort for better visualization (optional, but good practice for bar plots)
state_mean_rainfall = state_mean_rainfall.sort_values(by='ANNUAL', ascending=False)
```

```python
# Display the DataFrame with state-wise mean rainfall
print("State-wise Mean Annual Rainfall:")
display(state_mean_rainfall.head())
```

```python
# Create the bar graph
plt.figure(figsize=(15, 8))
sns.barplot(x='STATE_UT_NAME', y='ANNUAL', data=state_mean_rainfall, palette='viridis')
plt.title('Average Annual Rainfall by State/UT')
plt.xlabel('State/UT Name')
plt.ylabel('Average Annual Rainfall')
plt.xticks(rotation=90) # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
# Group by STATE_UT_NAME and count the number of districts
district_counts_by_state = df.groupby('STATE_UT_NAME')['DISTRICT'].count().reset_index()

# Rename the count column for clarity
district_counts_by_state.rename(columns={'DISTRICT': 'NUMBER_OF_DISTRICTS'}, inplace=True)

# Sort the results in descending order by the number of districts
district_counts_by_state = district_counts_by_state.sort_values(by='NUMBER_OF_DISTRICTS', ascending=False)

print("Number of Districts per State/UT:")
display(district_counts_by_state.head())
```

```python
# Group by state and find the index of the maximum annual rainfall in each state
idx_max_rainfall = df.groupby('STATE_UT_NAME')['ANNUAL'].idxmax()

# Use these indices to get the rows corresponding to the districts with max rainfall
district_max_rainfall_per_state = df.loc[idx_max_rainfall, ['STATE_UT_NAME', 'DISTRICT', 'ANNUAL']]

# Sort the results by annual rainfall in descending order for better readability
district_max_rainfall_per_state = district_max_rainfall_per_state.sort_values(by='ANNUAL', ascending=False).reset_index(drop=True)

print("District with the most annual rainfall per State/UT:")
display(district_max_rainfall_per_state.head())
```
